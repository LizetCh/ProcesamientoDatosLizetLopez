```python
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python
data = {
    "Comisiones": [22,13,33,38,23,47,29,38,41,32,20,13,47,38,44,29,38,37,14,34,25,27,25,43,34],
    "Llamadas":   [139,132,144,142,142,142,138,139,144,134,135,137,146,146,144,147,144,149,131,144,132,132,127,154,147],
    "Millas":     [2371,2226,2731,3351,2289,3449,3114,3342,2842,2625,2121,2219,3463,3290,3103,2122,2791,3209,2287,2848,2690,2933,2671,2988,2829]
}
```

```python
df = pd.DataFrame(data)
```

```python
# Gráfica
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.scatter(df["Llamadas"], df["Millas"], df["Comisiones"])
ax.set_xlabel("Llamadas de ventas")
ax.set_ylabel("Millas recorridas")
ax.set_zlabel("Comisiones (miles de dólares)")
ax.set_title("Gráfica de correlación múltiple")
plt.show()
```

## Análisis
La gráfica de correlación múltiple muestra que las comisiones de los agentes aumentan conforme incrementan tanto las millas recorridas como el número de llamadas de ventas.
