```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = {
    "Habitaciones_ocupadas": [23,47,21,39,37,29,23,44,45,16,30,42,54,27,34,15,19,38,44,47,43,38,51,61,39],
    "Ingreso": [1452,1361,1426,1470,1456,1430,1354,1442,1394,1459,1399,1458,1537,1425,1445,1439,1348,1450,1431,1446,1485,1405,1461,1490,1426]
}
```

```python
df = pd.DataFrame(data)
```

```python
# Grafica
plt.figure()
plt.scatter(df["Habitaciones_ocupadas"], df["Ingreso"])
plt.xlabel("Habitaciones ocupadas")
plt.ylabel("Ingreso del restaurante")
plt.title("Gráfica de correlación lineal simple")
plt.show()
```

## Análisis
La gráfica muestra una pequeña correlación con tendencia positiva pero débil.

## Prueba

```python
df.corr()
```

El coeficiente de correlación (r ≈ 0.423) confirma que existe una correlación lineal positiva, aunque no fuerte, lo que indica que el ingreso del restaurante aumenta con la ocupación hotelera, pero también depende de otros factores.
