```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
```

```python
#importar datos
df = pd.read_csv('wbes_sintetico.csv')
df.head()
```

## 1. Deleting extra columns

```python
df_copy = df.drop(columns=['exporter', 'digital_payments', 'sector', 'labor_productivity_usd_per_employee', 'successful'])
df_copy.head()
```

## 2. Checking for nulls and cleaning

```python
df_copy.isna().sum()
```

```python
df_clean = df_copy.dropna()
df_clean.isna().sum()
```

## 3. Variables independientes y variable dependiente

```python
X = df_clean.drop(columns=['sales_annual_usd'])
Y = df_clean[['sales_annual_usd']] 

X_encoded = pd.get_dummies(
    X,
    columns=['competition_level'],
    drop_first=True
)
```

```python
X_encoded
```

## 4. Modelo

```python
#Ajustar el modelo de regresión múltiple
model = LinearRegression() 
model.fit(X_encoded, Y) #Entrenando el modelo 
y_pred = model.predict(X_encoded)
#y_pred
```

## 5. Métricas

```python
#Metricas estadísticas
mse = mean_squared_error(Y, y_pred) #Error de la media cuadrática
r2 = r2_score(Y, y_pred) #varianza
```

```python
print('Valor de intersección : ', model.intercept_)
print('Coeficientes: ', model.coef_)
print('Error cuadrático medio', mse)
print('Varianza: ', r2)
```

## 6. Conclusiones
* Cada empleado adicional aumenta las ventas anuales en 1157 USD aprox
* Cada año adicional de operación aumenta ventas en 603 USD aprox
* Empresas con capacitación venden 6527 USD aprox. más que las que no
* Acceso a financiamiento está fuertemente asociado a mayores ventas
* Baja competencia = mayores ventas, comparadas con Alta competencia
* Competencia media = más ventas que alta, pero menos que baja

Los resultados del modelo son consistentes con la teoría económica, mostrando que el tamaño de la empresa, la experiencia, el acceso a financiamiento y un menor nivel de competencia están asociados positivamente con mayores ventas anuales.
